```python
import pandas as pd

real = pd.read_csv("../data/raw/train_transaction.csv")

synthetic = pd.read_parquet(
    "../data/synthetic/transactions.parquet"
)
```

```python
print("Real fraud rate:")
print(real["isFraud"].mean())

print("\nSynthetic fraud rate:")
print(synthetic["is_fraud"].mean())
```

```python
print("REAL")
print(real["ProductCD"].value_counts(normalize=True).sort_index())

print("\nSYNTHETIC")
print(synthetic["product_type"].value_counts(normalize=True).sort_index())
```

```python
print("REAL")
print(
    real.groupby("ProductCD")["isFraud"]
    .mean()
    .sort_values(ascending=False)
)

print("\nSYNTHETIC")
print(
    synthetic.groupby("product_type")["is_fraud"]
    .mean()
    .sort_values(ascending=False)
)
```

```python
print("REAL")
print(
    real.groupby("card6")["isFraud"]
    .agg(["count", "mean"])
)

print("\nSYNTHETIC")
print(
    synthetic.groupby("payment_type")["is_fraud"]
    .agg(["count", "mean"])
)
```

```python
percentiles = [0.50, 0.75, 0.90, 0.95, 0.99]

print("REAL")
print(
    real["TransactionAmt"].quantile(percentiles)
)

print("\nSYNTHETIC")
print(
    synthetic["amount"].quantile(percentiles)
)
```

```python
print("REAL")
print(
    real.groupby(
        real["TransactionID"].isin(
            set(
                pd.read_csv(
                    "../data/raw/train_identity.csv",
                    usecols=["TransactionID"]
                )["TransactionID"]
            )
        )
    )["isFraud"].mean()
)

print("\nSYNTHETIC")
print(
    synthetic.groupby("has_identity")["is_fraud"].mean()
)
```
